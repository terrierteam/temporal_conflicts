```python
from datasets import load_from_disk, load_dataset, Dataset
import pandas as pd
```

```python
templates_ds = load_dataset("coastalcph/fm_templates")["train"]
templates_ds
```

```python
templates_ds[0]
```

```python
dqa_ds = load_from_disk("dataset/DQA_statements_std_dedup")
dqa_ds
```

```python
df = dqa_ds.to_pandas()

# Group by matched_template and collect unique relations
unique_relations = df.groupby("matched_template")["relation"].unique().to_dict()
```

```python
unique_relations
```

```python
df_questions = pd.read_csv("dataset/dynamicqa_converted_templates.csv")

# Group by 'question' and aggregate the 'statement' values into lists
question_dict = df_questions.groupby('question')['statement'].apply(list).to_dict()

question_dict
```

```python
unique_relations_keys = set(unique_relations.keys())
unique_relations_keys
```

```python
question_dict_keys = set(question_dict.keys())
question_dict_keys
```

```python
unique_relations_keys == question_dict_keys
```

```python
templates_dict = {}
for q in unique_relations:
    for rel in unique_relations[q]:
        templates = question_dict[q]
        templates_dict[rel] = {"templates": templates}
templates_dict
```

```python
hf_dataset = Dataset.from_dict({k: [v] for k, v in templates_dict.items()})
hf_dataset
```

```python
hf_dataset[0]
```

```python
hf_dataset.save_to_disk("dataset/dqa_templates")
```
